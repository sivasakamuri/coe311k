```python
import numpy as np
import math
import matplotlib.pyplot as plt



def taylor_sin(x, terms): #function for taylor series for sin
    total = 0.0
    for n in range(terms):
        total += ((-1)**n * x**(2*n + 1)) / math.factorial(2*n + 1)
    return total



sin = np.sin(x) #Establsihing true sin

x = np.arange(0, 10.1, 0.1) #x from 0 to 10 in steps of 0.1

term = [1, 2, 3, 4] #Different terms to truncate with

errors = {} #Chat GPT used to understand how to use dictionares for an array of arrays



#Approximations and errors
for terms in term:
    approx = np.array([taylor_sin(x, terms) for x in x])
    errors[terms] = np.abs(approx - sin)

#Plot commands
plt.figure()
for terms in term:
    plt.plot(x, errors[terms], label=f"{terms} terms") # Chat GPT used to add legend, label, and grid lines

plt.xlabel("x")
plt.ylabel(" Error")
plt.title("Sin(x) Taylor Series Truncation Error")
plt.grid(True)
plt.legend()
plt.show()

#Discuss: As you increase terms it is defitnely true that the error gets smaller, causing the approximation to be more precise.
# However, as x gets bigger the error is also getting bigger. You would need more terms to help reduce the error so at these relatively small number of terms the error seems to be larger as the terms increase.
```
