```python
import random
import numpy
import matplotlib.pyplot as plt

def montecarlo_pi(number): #defining the actual monetecarlo randomizing function using random dots, a unit circle, and a square area
    count = 0
    for i in range(0,number):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        
        if (x*x + y*y < 1): #circle formula
            count = count + 1
    return 4*count/number

Pi = numpy.pi

samples = [] #arrays for data collection
errors = []

for n in range(10, 1001, 20): #GPT utilized to figure out how to get that final 1000 data set and how to add the data to the array without errasing the old data
    calcPi = montecarlo_pi(n)
    true_error = abs(Pi - calcPi)
    samples.append(n)
    errors.append(true_error)

plt.figure()
plt.plot(samples, errors)
plt.xlabel("Number of Samples")
plt.ylabel("Absolute Error")
plt.title("Monte Carlo Pi Absolute Error")
plt.show()
          
print("Gradual Error (15 decemial places): ", errors)
```
