```python
import numpy as np
import matplotlib.pyplot as plt

# lambda from dy/dx = -2y cos(x)
def lam(x):
    return -2 * np.cos(x)

# stability factor for Euler
def stability_factor(h, x):
    return 1 + h * lam(x)

# x range which is like HW7
x_vals = np.linspace(0, 10*np.pi, 1000)

# given step sizes
h_values = [0.01, 0.1, 0.25, 0.5, 1.0]

plt.figure()

# loop for each step size
for h in h_values:
    S = stability_factor(h, x_vals)
    
    # take magnitude, needed Chat GPT help to implement calculation
    max_stability = np.max(np.abs(S))
    
    print(f"h = {h:.2f}, max |1 + hλ(x)| = {max_stability:.3f}")
    
    # plot stability
    plt.plot(x_vals, np.abs(S), label=f"h={h}")

# reference stability boundary line, needed chat GPT to confrim stability understanding
plt.axhline(1, color='black', linestyle='--')

plt.xlabel("x")
plt.ylabel("|1 + hλ(x)|")
plt.title("Euler Stability Factor for dy/dx = -2y cos(x)")
plt.legend()
plt.show()

#This code analyzes the stability of Euler’s method by evaluating the magnitude of the given stability factor
#(|1 + h\lambda(x)|) across the interval.This codethis one predicts when the method will be stable by checking 
#if the magnitude stays below or equal to 1. As h increases, the data shows the stability factor exceeding 1 more often, 
#indicating instability. This helps show why larger step sizes led to higher errors and divergence.
```
