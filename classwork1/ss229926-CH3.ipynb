```python
import random
import numpy
import matplotlib.pyplot as plt

def montecarlo_pi(number): #defining the actual monetecarlo randomizing function using random dots, a unit circle, and a square area
    count = 0
    for i in range(0,number):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        
        if (x*x + y*y < 1): #circle formula
            count = count + 1
    return 4*count/number

Pi = numpy.pi

samples = [] #arrays for data collection
errors1 = []
errors2 = []



for i in range(100):                
    calcPi = montecarlo_pi(1000)     
    true_error = abs(Pi - calcPi)
    relative_error = abs((Pi - calcPi) / Pi)

    samples.append(i + 1) #GPT used to properly understand how to do the individual 100 times 1000 samples
    errors1.append(true_error)
    errors2.append(relative_error)


    
plt.figure()
    
plt.subplot(2, 1, 1) #absolute error sub plot 
plt.plot(samples, errors1, 'b')
plt.xlabel("Run Number")
plt.ylabel("Absolute Error")
plt.title("Monte Carlo Pi Absolute Error")

plt.subplot(2, 1, 2) #relative error sub plot 
plt.plot(samples, errors2, 'r')
plt.xlabel("Run Number")
plt.ylabel("Relative Error")
plt.title("Monte Carlo Pi Relative Error")

plt.tight_layout() #GPT used to figure out how to fix overlapping graphs
plt.show()
  
    
print("Error Data (15 decemial places): ", errors1, errors2)
```
