```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 2*np.pi, 50)


plt.subplot(3, 2, 1)
plt.plot(x, np.sin(x), 'om')
plt.title('Sin Wave')

plt.subplot(3, 2, 2)
plt.plot(x, np.cos(x), 'sb')
plt.title('Cos Wave')

plt.subplot(3, 2, 5)
k = np.random.rand(1000)
z = np.random.rand(1000)
size = np.random.rand(1000) * 50
color = np.random.rand(1000)
plt.scatter(k, z, s=size, c=color)
plt.title('Scatter Test')

plt.subplot(3, 2, 6)
plt.hist(x, 30, edgecolor='r')
plt.title('Histogram Test')

plt.show()
```
