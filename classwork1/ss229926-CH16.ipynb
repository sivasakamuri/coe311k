```python
import numpy as np
import matplotlib.pyplot as plt

# Function representing dy/dx = -2y
def f(x, y):
    return -2 * y

# Real solution
def real_solution(x):
    return np.exp(-2 * x)

# Euler method as a function
def euler_method(h):
    x_start, x_end = 0, 10
    num_steps = int((x_end - x_start) / h)

    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)

    # Initial condition
    y_values[0] = 1

    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])

    return x_values, y_values

# Step sizes
h_values = [0.1, 0.5, 1, 1.5]

plt.figure()

for h in h_values: #Chat gpt used to confirm the various plots
    x_vals, y_vals = euler_method(h)

    # Real values at same x
    y_real = real_solution(x_vals)

    # Local error
    local_error = np.abs(y_vals - y_real)

    # Global error
    global_error = np.max(local_error)

    print(f"h = {h}, Global Error = {global_error}")

    # Plot
    plt.plot(x_vals, local_error, label=f"h={h}")

plt.xlabel('x')
plt.ylabel('Error')
plt.title("Local Error for Euler Method")
plt.legend()
plt.show()
```
