```python
import numpy as np
import matplotlib.pyplot as plt

# Step sizes
h_values = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]


# Define the parameters
k = 1.0
x_start, x_end = 0, 10
y_0 = 1.0
v_0 = 0.0

plt.figure()

# Runs Eulers method for various step sizes
for h in h_values:

    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))

    y_values[0] = y_0
    v_values[0] = v_0

    for i in range(1, len(x_values)):
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]
# Set up for double derivitive 
        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)

    plt.plot(x_values, y_values, label=f"h={h}")

plt.xlabel('x')
plt.ylabel('y')
plt.title("Euler Method for d²y/dx² = -ky")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left') #Chat GPT used for formatting
plt.show()


# Euler’s method becomes unstable around larger step size where the solution will begin to grow
# Increasing K looks to increase oscillation frequency
# As h increases the function tends to diverge
# With K=10 and h=0.01 it oscilates at a high frequency
# Eulers method would not be able to solve this with a larger h 
```
