```python
import numpy as np
import matplotlib.pyplot as plt


# Define parameters
h = 0.1  # step size
x_start, x_end = 0, 2
num_steps = int((x_end - x_start) / h)

x_values = np.linspace(x_start, x_end, num_steps + 1)
y_values = np.zeros(num_steps + 1)


# Initial condition
y_values[0] = 1  # y(0) = 1

# Function representing dy/dx = -2y
def f(x, y):
    return -2 * y

# Implement Euler's method,needed to confirm with Chat GPT
for i in range(num_steps):
    y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])


# Exact solution of graph for comparison
exact_graph = np.exp(-2 * x_values)


# Plot the results
plt.plot(x_values, y_values, label="Euler's Method Approximation")
plt.plot(x_values, exact_graph, 'r--', label="Exact Graph")

plt.xlabel('x')
plt.ylabel('y')
plt.title("Approximation of dy/dx = -2y using Euler's Method")
plt.legend()
plt.show()
```
