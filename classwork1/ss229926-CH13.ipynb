```python
import numpy as np
import matplotlib.pyplot as plt
import time


#Naive matrix multiplication code from power point slides
def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    
    return C


#3x3 matrix to start and loops and size doubles each time
size = [3]
for _ in range(7):
    size.append(size[-1] * 2)

    
#empty arrays to store the time
naive_times = []
numpy_times = []




for n in size:
    print(f"Calculating {n}x{n} matrices...")
    
    # Random matrices
    A = np.random.rand(n, n)
    B = np.random.rand(n, n)
    
    # Convert to lists for naive version
    A_list = A.tolist()
    B_list = B.tolist()
    
    #ChatGPT utilized for some of the time tracking code
    #Starts the clock runs the function and then stops the time and logs that time
    
    #naive time
    start = time.time()
    naive_matrix_multiplication(A_list, B_list)
    end = time.time()
    naive_times.append(end - start)
    
    #numpy time
    start = time.time()
    np.dot(A, B)
    end = time.time()
    numpy_times.append(end - start)
    

#An organized chart with 3 columns showing the run time for each method as the size of the matrix increases
print(f"{'Matrix Size':>12} | {'Naive (s)':>12} | {'NumPy (s)':>12}")
print("-" * 45)

for i in range(len(size)):
    print(f"{size[i]:>12} | {naive_times[i]:>12f} | {numpy_times[i]:>12f}")
```
