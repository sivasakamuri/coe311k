```python
import numpy as np
import matplotlib.pyplot as plt

# function: dy/dt = -1000y + 3000 - 2000e^(-t)
def f(t, y):
    return -1000*y + 3000 - 2000*np.exp(-t)

# given analytical solution
def y_exact(t):
    return 3 - 0.998*np.exp(-1000*t) - 2.002*np.exp(-t)

# Euler Forward Method
def euler_forward(h, t0, tf, y0):
    n_steps = int((tf - t0) / h)
    t = np.linspace(t0, tf, n_steps + 1)
    y = np.zeros(n_steps + 1)
    
    y[0] = y0
    
    for i in range(n_steps):
        y[i+1] = y[i] + h * f(t[i], y[i])
    
    return t, y

# Parameters created from notes and chat GPT assisted
t0 = 0
tf = 0.01          #time period
y0 = 0
h = 1e-5           # step size

# Solving
t_euler, y_euler = euler_forward(h, t0, tf, y0)

t_fine = np.linspace(t0, tf, 1000)
y_true = y_exact(t_fine)


# Plotting
plt.figure(figsize=(8,5))
plt.plot(t_fine, y_true, label="Analytical Solution", linewidth=2)
plt.plot(t_euler, y_euler, '--', label="Euler Forward", linewidth=1)

plt.xlabel("t")
plt.ylabel("y(t)")
plt.title("Euler Forward vs Analytical Solution (Stiff)")
plt.legend()
plt.grid(True)
plt.show()
```
