```python
import numpy as np
import matplotlib.pyplot as plt


# Define function
f = lambda x: np.sin(x)

a, b = 0, 2 * np.pi # Interval
N = 50 # Number of sample points

# Creating a uniform grid
x = np.linspace(a, b, N)

# Step size
h = (b - a) / (N - 1)



# Sin derivative equals Cos
cos_graph = np.cos(x)

# Central Difference Approximation
central_diff = (f(x + h) - f(x - h)) / (2 * h)

# Forward Difference Approximation
forward_diff = (f(x + h) - f(x)) / h

# Backward Difference Approximation
backward_diff = (f(x) - f(x - h)) / h



# 4 subplots, Chat GPT used for remembering how to implement subplots cleanly
plt.figure(figsize=(12, 8))

# True Cos Plot
plt.subplot(2, 2, 1)
plt.plot(x, cos_graph, label='True cos(x)')
plt.title('Exact Derivative')
plt.legend()

# Forward Difference Plot
plt.subplot(2, 2, 2)
plt.plot(x, forward_diff, 'm', label='Forward Difference')
plt.title('Forward Difference Approx')
plt.legend()

# Backward Difference Plot
plt.subplot(2, 2, 3)
plt.plot(x, backward_diff, 'g', label='Backward Difference')
plt.title('Backward Difference Approx')
plt.legend()

# Central Difference Plot
plt.subplot(2, 2, 4)
plt.plot(x, central_diff, 'r', label='Central Difference')
plt.title('Central Difference Approx')
plt.legend()

plt.tight_layout()
plt.show()
```
