```python
import numpy as np
import matplotlib.pyplot as plt

# Compute x from 0 to 1 with steps of 0.01
X = np.arange(0, 1.01, 0.01)

# My chosen linear equation
my_m = 2.0
my_c = 1.0

# Adding in some random perturbation
perturbation = np.random.normal(0, 0.2, len(X))
Y = my_m * X + my_c + perturbation


#Using Least Sqaure to fit line
A = np.vstack([X, np.ones(len(X))]).T
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
Ypredictions = m*X + c


# Mean Squared Error (MSE) pulled from the slides
mse_ls = np.mean((Y - Ypredictions)**2)
print(f'Least Squares Mean Squared Error: {mse_ls:.4f}')


# R-squared value calculations pulled from the slides
ss_total = np.sum((Y - np.mean(Y))**2)

ss_res_ls = np.sum((Y - Ypredictions)**2)
r_squared_ls = 1 - (ss_res_ls / ss_total)
print(f'Least Squares R-squared: {r_squared_ls:.4f}')



# crank the polynomial up 1 degree at a time until it blatantly overfits
# crank the polynomial down 1 dress at a time (keep it > 0) until it underfits

for degree in [1,2,3,4,5,6,7,8]:

    coefficients = np.polyfit(X, Y, degree)
    polynomial = np.poly1d(coefficients)

    Y_poly_pred = polynomial(X)

    # Mean Squared Error (MSE) pulled from the slides
    mse_poly = np.mean((Y - Y_poly_pred)**2)

    # R-squared value calculations pulled from the slides
    ss_res_poly = np.sum((Y - Y_poly_pred)**2)
    r_squared_poly = 1 - (ss_res_poly / ss_total)

    print(f"\nPolynomial Degree {degree}")
    print(f'Mean Squared Error: {mse_poly:.4f}')
    print(f'R-squared: {r_squared_poly:.4f}')

    

print(f"\nChosen Slope: {my_m}, Chosen Intercept: {my_c}")
print(f"Estimated Slope: {m}, Estimated Intercept: {c}")


# Plot
plt.scatter(X, Y, s=10, label='Random Perturbation')

plt.plot(X, my_m*X + my_c, 'g', label='My Chosen Line')
plt.plot(X, Ypredictions, 'r', label='Least Squares Fitted Line')
plt.plot(X, Y_poly_pred, 'b', label='Polynomial Fit')

plt.legend(fontsize=7)
plt.xlabel('X')
plt.ylabel('Y')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Compute x from 0 to 1 with steps of 0.01
X = np.arange(0, 1.01, 0.01)

# My chosen linear equation
my_m = 2.0
my_c = 1.0

# Adding in some random perturbation
perturbation = np.random.normal(0, 0.2, len(X))
Y = my_m * X + my_c + perturbation


# Using Vandermonde Matrix to fit polynomial
degree = 3
V = np.vander(X, N=degree+1, increasing=True)
coefficients = np.linalg.lstsq(V, Y, rcond=None)[0]
Y_vander_pred = V @ coefficients


# Mean Squared Error (MSE) pulled from the slides
mse_vander = np.mean((Y - Y_vander_pred)**2)
print(f'Vandermonde Mean Squared Error: {mse_vander:.4f}')


# R-squared value calculations pulled from the slides
ss_total = np.sum((Y - np.mean(Y))**2)
ss_res_vander = np.sum((Y - Y_vander_pred)**2)
r_squared_vander = 1 - (ss_res_vander / ss_total)
print(f'Vandermonde R-squared: {r_squared_vander:.4f}')


print(f"\nChosen Slope: {my_m}, Chosen Intercept: {my_c}")


# Plot
plt.scatter(X, Y, s=10, label='Random Perturbation')

plt.plot(X, Y_vander_pred, 'b', label='Vandermonde Polynomial Fit')

plt.legend(fontsize=7)
plt.xlabel('X')
plt.ylabel('Y')

plt.show()
```
