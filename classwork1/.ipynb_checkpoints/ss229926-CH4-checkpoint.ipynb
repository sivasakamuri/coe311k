```python
import numpy as np

Pi = np.pi #true pi


def leibniz(n): 
    total = 0.0
    for k in range(1, n + 1):
        total += 4 * ((-1)**(k - 1) / (2*k - 1)) #leib formula
    return total
      


for n in range(100, 5001, 100): #100-5000 in steps of 100
    pi_leib = leibniz(n)
    relative_error = abs(Pi - pi_leib) / Pi
    print("Term:", n,", Relative Error:", relative_error)

```
